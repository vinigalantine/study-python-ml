## Building the Tokenizer

````python
class CharTokenizer:
    def __init__(self, vocabulary):
        self.token_id_for_char = {char: token_id for token_id, char in enumerate(vocabulary)}
        self.char_for_token_id = {token_id: char for token_id, char in enumerate(vocabulary)}
        
    @staticmethod
    def train_from_text(text):
        vocabulary = set(text)
        return CharTokenizer(sorted(list(vocabulary)))
    
    def encode(self, text):
        token_ids = []
        for char in text:
            token_ids.append(self.token_id_for_char[char])
        
        return torch.tensor(token_ids, dtype=torch.long)
    
    def decode(self, token_ids):
        chars = []
        for token_id in token_ids.tolist():
            chars.append(self.char_for_token_id[token_id])
        
        return ''.join(chars)
    
    def vocabulary_size(self):
        return len(self.token_id_for_char)
````

## Tesing the Tokenizer

````python
tokenizer = CharTokenizer.train_from_text(text)

print(f"Test encode: {tokenizer.encode("Hello world")}")
print(f"Test decode: {tokenizer.decode(tokenizer.encode("Hello world"))}")

print(f"Vocab size: {tokenizer.vocabulary_size()}")
````

## Softmax

### Why Use Softmax?

To create a probability distribution from logits, two conditions must be met:

1. **Range**: Every logit value must be converted into a value between 0 and 1.
2. **Sum**: The sum of all resulting values must equal one.

### How Does Softmax Work?

1. **Exponentiation**:
   - For each logit value, we calculate $e^{\text{logit}}$.
   - This ensures all output values are positive since raising $e$ (a positive number) to any power yields a positive result.
2. **Normalization**:
   - **Sum Calculation**: Compute the sum of all exponentiated values.
   - **Division**: Divide each exponentiated value by the sum to ensure that all outputs sum to one.

### Mathematical Representation

$$
\text{softmax}(x_i) = \frac{e^{x_i}}{\sum_j e^{x_j}}
$$

| Symbol | Meaning |
|---|---|
| $x_i$ | The logit (raw model output) for class/token $i$ |
| $e^{x_i}$ | Exponentiation of that logit — always positive, and larger logits grow disproportionately larger |
| $\sum_j e^{x_j}$ | The sum of exponentiated logits across every class/token $j$, used to normalize |
| $\text{softmax}(x_i)$ | The resulting probability for class/token $i$, in the range $(0, 1)$ |

**How it works**: exponentiating removes negative values (since $e^x > 0$ for any real $x$), and dividing by the total sum rescales everything so it adds up to exactly 1 — turning arbitrary logits into a valid probability distribution. The result is often visualized using a histogram.

**Real-life analogy**: think of splitting a pizza among friends based on how loudly each one asks for a slice. Exponentiation makes sure every "ask" counts as a positive amount (no one gets a negative slice), and dividing by the total "loudness" turns each person's share into a fraction of the whole pizza — the fractions always add up to 1 whole pizza.

### Properties of Softmax

- **Dominant Value Effect**: If one input value is significantly larger than the others, softmax assigns a high probability to this value, while the others receive probabilities close to zero.
- **Similarity Handling**: If multiple logits have similar values, softmax distributes probabilities more evenly among them.


# Calculating Attention

## Understanding the Attention Mechanism in Transformers

The **attention mechanism** is crucial for refining the embedding vectors for each token using the context of other embedding vectors — it lets every token "look at" every other token and decide how much each one matters.

## Overview of the Attention Block

- **Attention Block Function**: Computes attention scores and updates embedding values using these scores.
- **Components**: The attention block involves several steps, including calculating query, key, and value matrices and updating embedding vectors.

### Step-by-Step Implementation

1. **Query Matrix (Q)**
   - **Calculation**: Multiply the matrix of embedding vectors by a learned weight matrix $W_Q$ to produce the query matrix $Q$.
   - **Notation**: Represented as $Q$, often visualized as an array of vectors — one query vector per token.
2. **Key Matrix (K)**
   - **Calculation**: Similar to $Q$, the key matrix $K$ is obtained by multiplying the embedding vectors by matrix $W_K$.
   - **Concept**: Queries can be seen as questions a token asks ("what am I looking for?"), while keys advertise what each token offers ("what do I contain?").
3. **Computing Attention Scores**
   - **Similarity Check**: Calculate the dot product between vectors in the $Q$ and $K$ matrices to determine similarity — a high dot product means a query and key are well aligned.
   - **Matrix Multiplication**: Multiply the $Q$ matrix by the transposed $K$ matrix ($QK^T$) to get the attention scores for every token pair at once.
4. **Normalization with Softmax**
   - **Purpose**: Ensures all values are between 0 and 1 and sum to one, turning raw similarity scores into attention *weights*.
   - **Application**: Apply the softmax function to each row of the resulting matrix so each query's attention weights sum to 1 (i.e., each token distributes exactly "100% of its attention" across all tokens, including itself).

### Value Matrix (V)

- **Derivation**: Similar to $Q$ and $K$, the value matrix $V$ is derived using another learned weight matrix $W_V$.
- **Updating Embeddings**: Multiply each vector in $V$ by the corresponding attention weights (one weighted sum per token) to produce updated embeddings — this is what actually "mixes in" context from other tokens.

### Final Diagram of the Attention Block

- **Complete Process**:
  1. Compute $Q$, $K$, and $V$ matrices from the input embeddings.
  2. Compute attention scores from $Q$ and $K$, then normalize them with softmax.
  3. Update embeddings by combining the softmax weights with $V$.

### Why Scale by $\sqrt{d_k}$?

The course text mentions dividing by the square root of the key dimension but doesn't explain why — it's worth calling out explicitly:

- As the dimension of $Q$ and $K$ vectors ($d_k$) grows, dot products tend to grow larger in magnitude purely due to having more terms summed together.
- Very large values pushed into softmax produce an extremely "peaked" distribution (one weight near 1, the rest near 0), which makes gradients vanish during training and makes attention effectively ignore most tokens.
- Dividing by $\sqrt{d_k}$ rescales the scores back down to a more stable range before the softmax is applied, keeping gradients well-behaved.

### Formula for Attention

From the "Attention Is All You Need" paper:

$$
\text{Attention}(Q, K, V) = \text{softmax}\left(\frac{QK^T}{\sqrt{d_k}}\right)V
$$

| Symbol | Meaning |
|---|---|
| $Q$ | Query matrix, one row per token: $Q = XW_Q$ |
| $K$ | Key matrix, one row per token: $K = XW_K$ |
| $V$ | Value matrix, one row per token: $V = XW_V$ |
| $X$ | Matrix of input embedding vectors (one row per token) |
| $W_Q, W_K, W_V$ | Learned weight matrices, trained along with the rest of the model |
| $d_k$ | Dimensionality of the key (and query) vectors |
| $QK^T$ | Raw similarity (attention) scores between every pair of tokens |
| $\text{softmax}(\cdot)$ | Converts each row of scores into attention weights between 0 and 1 that sum to 1 |

**How it works**: $QK^T$ compares every query to every key via dot products, producing a similarity score per token pair. Scaling by $\sqrt{d_k}$ keeps those scores from growing too large, softmax turns each row into a clean probability distribution over "how much attention to pay to each token," and multiplying by $V$ uses those weights to build a new embedding for each token as a weighted blend of every token's value vector.

**Real-life analogy**: imagine a group discussion where each person (token) asks a question (query), and everyone else holds up a sign describing what they know (key). You compare your question to everyone's sign to decide how relevant they are (the dot product), scale down so no single person dominates the conversation purely because of sign length, then decide — via softmax — what fraction of your attention to give each person. Finally, you update your own understanding (embedding) by blending in what each person actually said (value), weighted by how much attention you gave them.

### Multi-Head Attention

- **Purpose**: A single attention computation can only capture one kind of relationship between tokens at a time. Multiple attention heads allow the model to learn several different relationships in parallel (e.g., one head might track grammatical structure while another tracks topical similarity).
- **Process**: Each head has its own $W_Q$, $W_K$, $W_V$ matrices and performs the attention calculation independently, typically using a smaller slice of the embedding dimension per head. The outputs from all heads are concatenated and passed through one more learned weight matrix $W_O$ to produce the final updated embedding vectors.

Understanding the attention mechanism is key to grasping how Transformer models process and generate text, enabling them to focus on relevant parts of the input sequence when making predictions.


# Masked Multi-Head Attention

## Preventing Cheating in Transformer Training

When training Transformer models, we convert parts of the training dataset into multiple training examples. However, a problem arises because the model can "cheat" by looking ahead at the entire input when predicting the next word. To prevent this, we need to modify the attention scores to ensure the model cannot access future tokens.

## Solution: Masking the Attention Matrix

To stop the model from "cheating," we implement a masking strategy:

1. **Attention Matrix Adjustment**:
   - **Problem**: Without masking, the raw attention scores let every token attend to every other token, including tokens that come *after* it in the sequence.
   - **Solution**: Update the attention matrix to mask out attention scores for future positions before the softmax step.
2. **Masking Strategy**:
   - **Method**: Assign all values above the main diagonal of the $QK^T$ score matrix to negative infinity.
   - **Softmax Effect**: The softmax function converts these negative infinity values to zeros, effectively ignoring them — so each token's attention weights only cover itself and earlier tokens.
3. **Implementation**:
   - **Diagram Update**: In the multi-head attention diagram, the masking step is inserted after the $QK^T$ matrix multiplication (and the $\sqrt{d_k}$ scaling) but before applying softmax.
   - **Masked Multi-Head Attention**: This approach is widely used in Transformer models, including GPT models, and is what makes them *causal* / *autoregressive* (each token can only be influenced by what came before it).

By applying a triangular mask, we ensure that the model only considers current and previous tokens when making predictions, thus maintaining the integrity of the training process.

### Why Negative Infinity (and Not Zero)?

It's worth being explicit about why the mask uses $-\infty$ rather than simply zeroing out the disallowed scores directly:

- Softmax is computed as $e^{x_i} / \sum_j e^{x_j}$, so an input score of $-\infty$ gives $e^{-\infty} = 0$ after exponentiation — guaranteeing that position receives *exactly* zero attention weight.
- Setting the raw score to $0$ instead would not work: $e^{0} = 1$, which is a perfectly normal (nonzero) contribution to the softmax, so the "forbidden" token would still receive attention.

### Formula for Masked Attention

Building on the unmasked attention formula from the previous section:

$$
\text{MaskedAttention}(Q, K, V) = \text{softmax}\left(\frac{QK^T}{\sqrt{d_k}} + M\right)V
$$

| Symbol | Meaning |
|---|---|
| $Q, K, V$ | Query, key, and value matrices, as defined in the Calculating Attention section |
| $d_k$ | Dimensionality of the key (and query) vectors |
| $M$ | Mask matrix: $M_{ij} = 0$ if token $j$ is at or before position $i$ (allowed), and $M_{ij} = -\infty$ if token $j$ comes after position $i$ (forbidden) |
| $\text{MaskedAttention}(\cdot)$ | Same as regular attention, but each token's output only ever depends on itself and earlier tokens |

**How it works**: $M$ is an upper-triangular matrix of $-\infty$ values (everything above the main diagonal) added element-wise to the scaled scores. Adding $-\infty$ to a forbidden position's score drives it to $-\infty$ regardless of what $QK^T$ computed, so after softmax that position's weight collapses to exactly $0$ — while the allowed (on-diagonal and below) positions pass through unaffected and still sum to $1$ across each row.

**Real-life analogy**: imagine taking a test where you're only allowed to use notes from questions you've already answered — a sheet of paper is dropped over every future question so you physically cannot read it. The mask doesn't change how good your reasoning (the attention scores) is for the questions you *can* see; it just makes the forbidden ones completely inaccessible (probability zero), so your answer to question $i$ never leaks information from question $i+1$ or beyond.

### Connecting Back to What We've Already Built

This same upper-triangular masking idea was explored earlier using `torch.tril` (see the tril/causal-mask note in the PyTorch basics notebook) — in practice, the mask is usually built once as a fixed lower-triangular boolean matrix (`torch.tril`), and the forbidden (upper-triangular) positions are filled with `-inf` via `masked_fill` right before the softmax call in the attention block.


# Dropout

## What is a Dropout Layer?

- **Function in Neural Networks**: In a feedforward neural network, each output from a previous layer is connected to all neurons in the next layer.
- **Purpose of Dropout**: The Dropout layer introduces randomness by setting a certain percentage of the inputs to zero during training. This percentage is usually specified as a hyperparameter (the **dropout rate**, $p$).

## How Dropout Works

- **Random Selection**: On each training iteration, different inputs are randomly selected and set to zero. This list of "disabled" inputs changes with each iteration.
- **Generalization**: By doing this, the model does not become overly reliant on specific features, which helps it generalize better to new, unseen data. This process reduces the risk of overfitting.

### Keeping the Scale Consistent (Inverted Dropout)

The course text doesn't mention this, but it's important for understanding why dropout doesn't silently shrink the network's output:

- If we simply zeroed out a fraction $p$ of the values and did nothing else, the *sum* of values flowing into the next layer would shrink by roughly $p$ on average — a different "typical magnitude" than what the next layer sees at inference time (when dropout is off and nothing is zeroed).
- To fix this, PyTorch (and most modern frameworks) use **inverted dropout**: the surviving (non-zeroed) values are scaled up by $\frac{1}{1-p}$ during training, so the expected sum stays the same whether dropout is on or off.

$$
\text{output}_i = \begin{cases} 0 & \text{with probability } p \\ \dfrac{x_i}{1-p} & \text{with probability } 1-p \end{cases}
$$

| Symbol | Meaning |
|---|---|
| $x_i$ | The original (pre-dropout) value at position $i$ |
| $p$ | Dropout rate — probability that a given value is zeroed out |
| $1-p$ | Probability a value survives |
| $\dfrac{1}{1-p}$ | Scale-up factor applied to surviving values, keeping the expected magnitude unchanged |

**How it works**: zeroing happens independently and randomly for each value on every forward pass, and dividing the survivors by $1-p$ compensates for the values that got dropped — on average, $\text{output}_i$ has the same expected value as $x_i$, so no rescaling is needed later at inference time.

**Real-life analogy**: imagine a sports team practicing with a random subset of players benched every single practice (dropout), but each remaining player is asked to cover a bit more ground to compensate (the $\frac{1}{1-p}$ scaling) — so the team's overall effectiveness in practice still roughly matches what it will be on game day, when the full roster (no dropout) plays together.

## Training vs. Inference

- **During Training**: The Dropout layer is active, randomly disabling inputs to help generalization.
- **During Inference**: When the model is used for generating text or making predictions, the Dropout layer is disabled, ensuring that all features are fully utilized.
- **Switching Modes in PyTorch**: `nn.Dropout` automatically checks whether the model is in training or evaluation mode — calling `model.train()` turns dropout on, and `model.eval()` turns it off (no inputs are zeroed and no scaling is applied). It's easy to forget this switch and accidentally evaluate a model with dropout still active, which produces noisy, non-reproducible results.

## Implementation in PyTorch

- **Pre-built in PyTorch**: We do not need to implement the Dropout layer ourselves, as PyTorch provides a built-in implementation through the `nn.Dropout` class — it is typically inserted right after the attention output (and after the feed-forward block) in a Transformer block.

```python
dropout_rate = 0.1
dropout = nn.Dropout(dropout_rate)

sample_values = torch.ones(10)
print(dropout(sample_values))  # some entries become 0, survivors scaled by 1 / (1 - dropout_rate)
```

The Dropout technique is a crucial component in training robust neural networks, especially when dealing with complex data and models. It ensures that the model does not overfit and can generalize well across different datasets.


### Implementing the Attention Block

````python
config = {
    "vocabulary_size": tokenizer.vocabulary_size(),
    "context_size": 256,
    "embedding_dim": 768,
    "attention_heads": 12,
    "layers_num": 10,
    "dropout_rate": 0.1,
    "use_bias": False
}

# Each head only works with a slice of the embedding, so the heads' outputs can be concatenated back to embedding_dim
config["head_size"] = config["embedding_dim"] // config["attention_heads"]

class AttentionHead(nn.Module):
    def __init__(self, config):
        super().__init__()
        # Project embedding_dim -> head_size (not attention_heads!) so heads_num * head_size == embedding_dim
        self.Q_weights = nn.Linear(config["embedding_dim"], config["head_size"], config["use_bias"]) # Querry
        self.K_weights = nn.Linear(config["embedding_dim"], config["head_size"], config["use_bias"]) # Keys
        self.V_weights = nn.Linear(config["embedding_dim"], config["head_size"], config["use_bias"])
        
        self.dropout = nn.Dropout(config["dropout_rate"])
        
        # Fixed (non-learned) lower-triangular mask, registered as a buffer so it moves with .to(device) but isn't trained
        casual_attention_mask = torch.tril(torch.ones(config["context_size"], config["context_size"]))
        self.register_buffer('casual_attention_mask', casual_attention_mask)
        
    def forward(self, input): # input = (B, C, embedding_dim)
        batch_size, tokens_num, embedding_dim = input.shape
        Q = self.Q_weights(input) # (B, C, head_size)
        K = self.K_weights(input) # (B, C, head_size)
        V = self.V_weights(input) # (B, C, head_size)
        
        attention_scores = Q @ K.transpose(1, 2)
        # Slice the mask to the actual sequence length, since input may be shorter than context_size
        attention_scores = attention_scores.masked_fill(
            self.casual_attention_mask[:tokens_num,:tokens_num] == 0,
            -torch.inf
        )
        attention_scores = attention_scores / ( K.shape[-1] ** 0.5 ) # scale by sqrt(d_k) to keep softmax stable
        attention_scores = torch.softmax(attention_scores, dim=-1)
        attention_scores = self.dropout(attention_scores) # dropout applied to attention weights, not the output
        
        return attention_scores @ V
    
    
input = torch.rand(8, config["context_size"], config["embedding_dim"])

ah = AttentionHead(config)

output = ah(input)

print(output.shape) # torch.Size([8, 256, 64]) -> (B, C, head_size)

class MultiHeadAttention(nn.Module):
    def __init__(self, config):
        super().__init__()
        
        # One independent AttentionHead per head, each learning its own Q/K/V projections
        head_list = [AttentionHead(config) for _ in range(config["attention_heads"])]
        self.heads = nn.ModuleList(head_list)
        
        # Concatenated heads give back head_size * attention_heads == embedding_dim
        self.linear = nn.Linear(config["head_size"] * config["attention_heads"], config["embedding_dim"], config["use_bias"])
        self.dropout = nn.Dropout(config["dropout_rate"])
        
    def forward(self, input):
        heads_outputs = [head(input) for head in self.heads]
        
        # Concatenate along the last (feature) dimension, not dim=1 (that's the token/sequence dimension)
        scores_change = torch.cat(heads_outputs, dim=-1)
        
        scores_change = self.linear(scores_change)
        
        return self.dropout(scores_change) # dropout applied again, this time to the block's final output
    
mha = MultiHeadAttention(config)

output = mha(input)

print(output.shape) # torch.Size([8, 256, 768]) -> (B, C, embedding_dim)
````

# The Transformer Block

## Overview of a Transformer Block

- **Input Processing**: Each input token is converted into an embedding vector and passed to the multi-head attention block.
- **Feedforward Network**: The output from the multi-head attention block is sent to a feedforward network layer.
- **Chaining**: A Transformer model consists of multiple such blocks chained together.

## Feedforward Layer

- **Function**: Applied to each embedding vector individually (unlike attention, there's no mixing of information *between* tokens in this layer).
- **Process**:
  - **Linear Layer**: Each embedding vector is passed to a linear layer with an output size four times larger than the input size.
  - **Activation Function**: The GELU activation function is applied to introduce nonlinearity.
  - **Size Reduction**: The output vector is then reduced to the original embedding size.

### GELU Activation Function

- **Comparison**: GELU is similar to the ReLU activation function but has a smoother shape, which helps in achieving better training results.

$$
\text{ReLU}(x) = \max(0, x) \qquad \qquad \text{GELU}(x) \approx x \cdot \sigma(1.702x)
$$

| Symbol | Meaning |
|---|---|
| $x$ | The pre-activation input value |
| $\max(0, x)$ | ReLU: a hard cutoff — any negative input becomes exactly $0$, with a sharp "elbow" at $x=0$ |
| $\sigma(\cdot)$ | The sigmoid function, $\sigma(z) = 1/(1+e^{-z})$ |
| $\text{GELU}(x)$ | A smooth approximation — instead of a hard cutoff, it gradually scales $x$ down near zero, including small negative values passing through with a small negative output |

**How it works**: unlike ReLU's sharp corner, GELU transitions smoothly from "mostly blocked" to "mostly passed through," which keeps gradients well-behaved for inputs near zero and is one of the reasons it tends to train better in Transformers (used in BERT, GPT, etc.).

**Real-life analogy**: ReLU is like a strict bouncer who either lets you in fully or turns you away completely at the door (nothing in between). GELU is like a more lenient bouncer who lets most of you through if you look mostly fine, partially limits you if you're borderline, and only turns away those who are clearly not dressed for the occasion — a gradual rather than all-or-nothing decision.

## Additional Components

### Residual Connections

- **Purpose**: Helps with the vanishing gradient problem in deep networks.
- **Implementation**: The input to a block is added to the block's output before passing it forward, improving training efficiency.

$$
\text{output} = \text{SubLayer}(x) + x
$$

| Symbol | Meaning |
|---|---|
| $x$ | The input to the sub-layer (e.g., multi-head attention or the feedforward layer) |
| $\text{SubLayer}(x)$ | Whatever transformation that sub-layer applies |
| $+\, x$ | The original input, added back in unchanged — the "skip connection" |

**How it works**: during backpropagation, gradients can flow directly through the `+ x` path even if the gradient through `SubLayer(x)` shrinks toward zero, so deep stacks of Transformer blocks don't suffer as much from vanishing gradients. It also means each sub-layer only needs to learn a *residual* (a correction/refinement) on top of its input, rather than having to reconstruct the entire signal from scratch.

**Real-life analogy**: think of editing a draft document with track changes — instead of rewriting the whole document (`SubLayer(x)` replacing `x` entirely), you only add your suggested edits on top of the original text (`SubLayer(x) + x`), so the original content is never at risk of being lost, even if your edit for this pass is small or imperfect.

### Layer Normalization and Dropout

- **Layer Normalization**: Scales the output so that the mean becomes zero and variance becomes one, leading to faster and more stable training.
- **Dropout Layer**: Helps prevent overfitting by setting a portion of the output to zero during training (see the earlier Dropout section for how and why).

## Summary

### Components of a Transformer Block

1. **Multi-Head Attention**
2. **Feedforward Network**
3. **Layer Normalization**
4. **Dropout**
5. **Residual Connections**

These elements work together to process and refine the input data, making Transformer models highly effective for various NLP tasks.

Now that we've covered the theoretical aspects, let's proceed to implement the Transformer block in our model.


### Implementing Transform Block

````python
class FeedForward(nn.Module):
    def __init__(self, config):
        super().__init__()
        
        self.linear_layers = nn.Sequential(
            nn.Linear(config["embedding_dim"], config["embedding_dim"] * 4), # expand
            nn.GELU(),
            nn.Linear(config["embedding_dim"] * 4, config["embedding_dim"]), # project back down
            nn.Dropout(config["dropout_rate"]),
        )
    
    def forward(self, input): # applied independently per token, no mixing across the sequence dim
        return self.linear_layers(input)
    
ff = FeedForward(config)

input = torch.rand(8, config["context_size"], config["embedding_dim"])
output = ff(input)

print(output.shape) # torch.Size([8, 256, 768]) -> (B, C, embedding_dim), unchanged

class Block(nn.Module):
    def __init__(self, config):
        super().__init__()
        
        self.multi_head = MultiHeadAttention(config)
        self.layer_norm_1 = nn.LayerNorm(config["embedding_dim"])
        
        self.feed_forward = FeedForward(config)
        self.layer_norm_2 = nn.LayerNorm(config["embedding_dim"])
        
    def forward(self, input):
        residual = input
        x = self.multi_head(self.layer_norm_1(input)) # pre-norm: normalize before the sub-layer, not after
        x = x + residual # residual connection around attention
        
        residual = x
        x = self.feed_forward(self.layer_norm_2(x)) # must be feed_forward here, not multi_head again
        return x + residual # residual connection around the feedforward block
    

b = Block(config)

input = torch.rand(8, config["context_size"], config["embedding_dim"])

output = b(input)

print(output.shape) # torch.Size([8, 256, 768]) -> (B, C, embedding_dim), unchanged
````

## Feedforward Block Implementation

To define a feedforward block, we create a new class inheriting from `nn.Module`.

### Key Components

- **Linear Layers**: The first linear layer projects the input dimension to four times its size, followed by a GELU activation function and another linear layer that reduces the dimension back to the original size.
- **Dropout Layer**: Added to prevent overfitting.

### Testing the Feedforward Block

- **Instance Creation**: An instance of the feedforward block is created.
- **Input Testing**: A random input is passed through the block to check if the output dimensions are as expected — `torch.Size([8, 256, 768])`, unchanged from the input shape (the feedforward block only transforms the embedding dimension internally; it doesn't change the shape it outputs).

## Transformer Block Implementation

With both the multi-head attention and feedforward blocks ready, we can now implement the Transformer block.

### Transformer Block Components

- **Multi-Head Attention**: Uses the multi-head attention block.
- **Feedforward Block**: Incorporates the previously defined feedforward block.
- **Layer Normalization**: Two normalization layers are included to stabilize the training process — one before attention, one before the feedforward block.
- **Residual Connections**: Two residual connections are used to help with the vanishing gradient problem — one around attention, one around the feedforward block.

### A Bug Worth Flagging

While wiring this up, it's easy to accidentally reuse `self.multi_head` for the second sub-layer instead of `self.feed_forward` — both calls look structurally identical (`self.<something>(self.layer_norm_<n>(x))`), so this typo doesn't raise any shape errors; it silently runs attention twice instead of attention-then-feedforward. Always double-check that the second residual branch in `forward` calls `self.feed_forward`, not `self.multi_head`.

### Pre-Norm vs. Post-Norm

The course text doesn't name this explicitly, but it's worth knowing: normalizing *before* each sub-layer (`self.multi_head(self.layer_norm_1(input))`) rather than *after* it is called **pre-norm**, and it's the arrangement most modern Transformer implementations (including GPT-style models) use, because it tends to produce more stable gradients in very deep stacks of blocks compared to the original "post-norm" arrangement from the "Attention Is All You Need" paper.

### Testing the Transformer Block

- **Instance Creation**: Create an instance of the Transformer block.
- **Input Testing**: A random input is passed through the block to verify the output dimensions — `torch.Size([8, 256, 768])`, the same shape as the input. This is intentional: a Transformer block *refines* embedding vectors, it never changes their shape, which is exactly what allows multiple blocks to be chained together.

This implementation completes the core components of our Transformer model, allowing us to handle complex tasks involving sequential data.


# Putting It All Together

## Model Components Overview

This ties together everything built so far — tokenizer, embeddings, positional encoding, and the Transformer block — into the full end-to-end flow of the model.

1. **Tokenization and Embedding**:
   - **Input Splitting**: The input is split into tokens, such as words, subwords, or characters.
   - **Token ID Conversion**: These tokens are converted into token IDs.
   - **Embedding Vectors**: Token IDs are converted into embedding vectors.
   - **Positional Encoding**: Positional encoding values are added to the embedding vectors, resulting in input vectors for the model.
2. **Layer Processing**:
   - **Layer Passing**: The input vectors are passed through the layers of the model sequentially.
   - **Shape Consistency**: The output from each layer has the same shape as the input, but the values of embedding vectors evolve with each layer.
   - **Batch Processing**: During training, a batch of inputs is processed together to optimize resource usage.
3. **Prediction and Output**:
   - **Final Layer Output**: After the final layer, the output consists of a series of embedding vectors.
   - **Token Prediction**:
     - **Linear Layer**: Converts each embedding vector into a new vector with prediction scores for each token in the vocabulary. A score, such as 0.39, indicates the model's confidence in generating a token with a specific ID.
     - **Softmax Conversion**: Converts the prediction scores into probabilities using the softmax function. This results in a probability distribution for each token, where the sum of probabilities for each position equals one.

By following this structure, we ensure that the Transformer model processes and generates data accurately, leveraging the power of multi-head attention and feedforward networks. This design allows the model to handle complex tasks like language modeling, translation, and more.

The next step is to implement this entire structure in PyTorch, integrating all the components into a cohesive model.


### Implementing the Transformer Model

````python
class DemoGPT(nn.Module):
    def __init__(self, config):
        super().__init__()
        
        self.token_embedding_layer = nn.Embedding(config["vocabulary_size"], config["embedding_dim"])
        # Positional embeddings are looked up by position index (0..context_size-1), not by token identity
        self.positional_embedding_layer = nn.Embedding(config["context_size"], config["embedding_dim"])
        
        blocks = [Block(config) for _ in range(config["layers_num"])]
        self.layers = nn.Sequential(*blocks) # stacks layers_num Transformer blocks back-to-back
        
        self.layer_norm = nn.LayerNorm(config["embedding_dim"])
        # "Unembedding": projects embedding_dim back up to vocabulary_size, one score per possible next token
        self.unembedding = nn.Linear(config["embedding_dim"], config["vocabulary_size"], bias=False)
    
    def forward(self, token_ids): # token_ids = (B, C)
        batch_size, tokens_num = token_ids.shape
        
        x = self.token_embedding_layer(token_ids) # (B, C, embedding_dim)
        sequence = torch.arange(tokens_num, device=device)
        x = x + self.positional_embedding_layer(sequence) # broadcasts (C, embedding_dim) across the batch
        
        x = self.layers(x)
        x = self.layer_norm(x)
        x = self.unembedding(x) # (B, C, vocabulary_size) -> logits per token position
        
        return x
    
model = DemoGPT(config).to(device)

output = model(tokenizer.encode("Hi").unsqueeze(dim=0).to(device)) # unsqueeze adds the batch dim: (C,) -> (1, C)

print(output.shape) # torch.Size([1, 2, 65]) -> (B, C, vocabulary_size)

def generate(model, prompt_ids, max_tokens):
    output_ids = prompt_ids
    
    for _ in range(max_tokens):
        if output_ids.shape[1] >= config["context_size"]:
            break # stop once we'd exceed the model's max context length
        
        with torch.no_grad(): # inference only, no need to track gradients here
            logits = model(output_ids)
            
        logits = logits[:,-1, :] # only the last position's logits predict the next token
        probs = F.softmax(logits, dim=1)
        # Sample a random token given the softmax distribution
        next_token_id = torch.multinomial(probs, num_samples=1)
        #Add new token to the output, and repeat the process
        output_ids = torch.cat([output_ids, next_token_id], dim=1)
        
    return output_ids

def generate_with_prompt(model, tokenize, prompt, max_tokens=100):
    model.eval() # disables dropout for inference
    prompt = tokenizer.encode(prompt).unsqueeze(dim=0).to(device)
    
    return tokenizer.decode(generate(model, prompt, max_tokens=max_tokens)[0])

generate_with_prompt(model, tokenizer, "First Citizen:\n") # gibberish expected: the model hasn't been trained yet
````

## Building Blocks of the Model

### Token Embedding

- **Purpose**: Converts input token IDs into embedding vectors.
- **Implementation**: Using `nn.Embedding`, which maps each token ID to its corresponding embedding vector.
- **Parameters**:
  - Number of embedding vectors (vocabulary size).
  - Dimensionality of the embedding vectors.

### Positional Encoding

- **Purpose**: Adds positional information to the embedding vectors, helping the model understand the order of tokens.
- **Implementation**: Another `nn.Embedding` layer, but with the number of vectors equal to the context size (the maximum number of tokens in a sequence).

### Transformer Blocks

- **Creation**: We create multiple Transformer blocks, each containing a multi-head attention layer and a feedforward layer.
- **Combination**: These blocks are combined sequentially, allowing data to pass through each layer.

### Final Layer Normalization and Linear Layer

- **Layer Normalization**: Normalizes the output from the last Transformer block.
- **Linear Layer**: Converts the final output vectors into prediction scores for each token in the vocabulary, determining which token should be generated next. This final projection is sometimes called the **"unembedding"** or **"language modeling head"** — it mirrors the token embedding layer in reverse, going from `embedding_dim` back up to `vocabulary_size`, so its output size must always match the tokenizer's vocabulary (`tokenizer.vocabulary_size()`).

## Model Implementation in PyTorch

### DemoGPT Class

The `DemoGPT` class wires together everything built in this notebook so far — token embeddings, positional embeddings, the stack of Transformer blocks, and the final normalization + unembedding layer — into one end-to-end model.

The `unsqueeze` method is used to add an extra dimension for batch processing — since the model always expects a `(batch_size, tokens_num)` input, a single sequence needs a batch dimension of size 1 added before it can be passed in.
- **Expected Output**: `torch.Size([1, 2, 65])` — one batch, two tokens (`"H"` and `"i"`), and one logit per vocabulary entry (65 unique characters in this dataset) for each token position.

## Text Generation Functions

### `generate` Function

- **Purpose**: Generates text sequences using the model.
- **Process**:
  - Initializes output with prompt IDs.
  - Iteratively generates tokens using the model, sampling the next token based on the softmax probabilities.
- **Why Sample Instead of Always Picking the Top Token?**: Taking `torch.multinomial` of the probability distribution rather than always picking the highest-probability token (`argmax`) introduces randomness, which is what allows the same prompt to produce different completions on different runs — this is the same mechanism behind a model's "temperature" setting in many chat tools.

### `generate_with_prompt` Function

- **Purpose**: A wrapper for the `generate` function.
- **Functionality**: Takes a prompt string, encodes it into token IDs, passes it to the `generate` function, and decodes the output back into a readable string. Calling `model.eval()` first is important — without it, dropout would still be active and randomly zero out values during generation, which is a training-time behavior that has no business affecting inference.

## Training Workflow

### 1. Data Loading

- **Data Loader**: In each iteration, we use a data loader to load a batch of inputs and targets.
- **Inputs**: Tensor of token IDs representing the input text.
- **Targets**: Tensor of token IDs representing the expected output — for a language model, this is simply the input sequence shifted one token to the right, so predicting "the next token" at every position doubles as the training signal.

### 2. Model Prediction

- **Forward Pass**: The batch of inputs is passed through the model, which generates predictions.
- **Predictions**: The model outputs logits, which are raw prediction scores for each token in the vocabulary.

### 3. Loss Calculation

- **Loss Function**: We calculate the loss by comparing the model's predictions with the actual targets.
- **Purpose**: The loss function quantifies how far the model's predictions are from the actual values.
- **Connecting Back to Softmax**: The course text doesn't name it here, but for this kind of next-token prediction task the loss function is almost always **cross-entropy**, which is just $-\log(\text{softmax probability assigned to the correct token})$ — the softmax section covered earlier in this notebook is exactly what turns the model's logits into the probabilities this loss function needs. PyTorch's `nn.CrossEntropyLoss` conveniently applies softmax internally, so it's typically fed raw logits directly, not pre-softmaxed probabilities.

### 4. Backpropagation

- **Gradient Calculation**: With the calculated loss, we perform backpropagation to compute the gradients of the loss with respect to the model parameters.

### 5. Parameter Update

- **Optimizer**: Using the optimizer, we update the model's parameters to minimize the loss.
- **Iteration Completion**: This completes a single iteration of training.

### 6. Repeating the Process

- **Epochs**: The training process is repeated for many iterations and epochs, using different batches of data to improve the model's performance.

By following this workflow, the model learns to generate more accurate text outputs over time. Each batch of data helps the model fine-tune its weights and biases, gradually improving its predictions.


### Implementing the training

````python
batch_size = 64

train_interations = 5000
evaluation_interval = 100
learning_rate = 4e-4

train_data = tokenizer.encode(text).to(device) # 1D tensor of token ids for the whole dataset
train_dataset = TokenIdsDataset(train_data, config["context_size"])

from torch.utils.data import Dataset, DataLoader, RandomSampler

# replacement=True lets us draw more samples than exist in the dataset, one per training step
train_sampler = RandomSampler(train_dataset, num_samples=batch_size * train_interations, replacement=True)
train_dataloader = DataLoader(train_dataset, batch_size=batch_size, sampler=train_sampler)

optimizer = torch.optim.AdamW(model.parameters(), lr=learning_rate)

for step_num, sample in enumerate(train_dataloader): # was "enumarate" (typo), which raises NameError
    
    model.train() # re-enable dropout; generate_with_prompt below switches to eval()
    input, targets = sample
    logits = model(input)
    
    # flatten batch and sequence dims together, since cross_entropy expects (N, C) logits and (N,) targets
    logits_view = logits.view(batch_size * config["context_size"], config["vocabulary_size"])
    targets_view = targets.view(batch_size * config["context_size"])
    
    loss = F.cross_entropy(logits_view, targets_view)
    loss.backward()
    
    optimizer.step()
    optimizer.zero_grad(set_to_none=True) # frees gradient tensors instead of zeroing them, slightly faster
    
    print(f"Step {step_num} | Loss {loss.item():3f}")
    
    if step_num % evaluation_interval == 0:
        print("Demo GPT: \n" + generate_with_prompt(model, tokenizer, "\n"))
````
