# Learning Numpy

## Why should I use Numpy?

It is a package is for numerical computing, using C "under the hood" make operations faster than using plain python. Check the cases below:

```python
import time
import numpy as np
```

### Generating an array of 100000000 floats with numpy

```python
x = np.random.random(100000000)
```

### Case 1

Using plain python to calculate the mean of the array

```python
start = time.time()
sum(x) / len(x)
print("Operation time:",time.time() - start)
```

### Case 2

Usinf numpy to calculate the mean of the array

```python
start = time.time()
np.mean(x)
print("Operation time:", time.time() - start)
```

## Accessing, Deleting, and Inserting Elements Into ndarrays


Accessing elements on an array and altering element

```python
x = np.array([1,2,3,4,5])
print(x)

print('1st element:', x[0])
print('2nd element:', x[1])
print('3th element:', x[2])

# Using negative numbers
print('1st element:', x[-5])
print('2nd element:', x[-4])
print('5th element:', x[-1])

x[2] = 20
print(x)

```

### Manipulating arrays

Accessing and deleting

```python
x = np.arange(1,10).reshape((3,3))
print(x)
print(x[1,1])

x = np.array([1,2,3,4,5])
print(x)

# Deleting first and last elements
x = np.delete(x, [0,4])
print(x)

X = np.arange(1,10).reshape((3,3))
print('\n',X)

Y = np.delete(X, 0, axis=0)
print('\n',Y)

Z = np.delete(Y, [0,2], axis=1)
print('\n',Z)
```

Appending

```python
x = np.array([1,2,3,4,5])
print(x)

x = np.append(x, 6)
print(x)

x = np.append(x, [7,8,9])
print(x)

X = np.arange(1,10).reshape((3,3))
print(X)

Y = np.append(X, [[10,11,12]], axis=0)
print(Y)
```

Inserting

```python
x = np.array([1,2,5])
print('\n',x)

x = np.insert(x,2,[3,4])
print('\n',x)

X = np.array([[1,2,3],[7,8,9]])
print('\n',X)

Y = np.insert(X, 1, [4,5,6], axis=0)
print('\n',Y)
```

Stacking

```python
x = np.array([1,2])

Y = np.array([[3,4],[5,6]])

print(x,'\n',Y)

Z = np.vstack((x,Y))
print('\n',Z)

W = np.hstack((Y, x.reshape(2,1)))
print('\n',W)
```

Slicing

- ndarray[start:end]
- ndarray[start:]
- ndarray[:end]

```python
x = np.array([1,2,3,4,5])

print(x[1:3])
print(x[2:])
print(x[:2])

print("===================")

y = np.arange(1,21).reshape(4,5)
print(y)
print(y[1:4, 2:5])
```

Pay attention to this:

```python
X = np.arange(1,21).reshape(4,5)

print(X)
print("==========")
y = X[1:,2:]
print(y)
print("==========")

y[2,2] = 555;

print(y)
print('\n',X)
print("==========")

X = np.arange(1,21).reshape(4,5)
z = X[1:,2:].copy()

z[2,2] = 555;

print(z)
print('\n',X)
```

Boolean Indexing, Set Operations, and Sorting

```python
X = np.arange(25).reshape(5,5)

print(X[X > 10])
```

Comparison operations

```python
x = np.array([1,2,3,4,5])
y = np.array([6,7,2,8,4])

# We print x
print()
print('x = ', x)

# We print y
print()
print('y = ', y)

# We use set operations to compare x and y:
print()
print('The elements that are both in x and y:', np.intersect1d(x,y))
print('The elements that are in x that are not in y:', np.setdiff1d(x,y))
print('All the elements of x and y:',np.union1d(x,y))
```

Sorting

```python
x = np.random.randint(1,11,size=(10,))

print()
print('Original x = ', x)

print()
print('Sorted x (out of place):', np.sort(x))
print()
print('Sorted unique elements:', np.unique(x))

print()
print('x after sorting:', x)

X = np.random.randint(1,11,size=(5,5))

print()
print('Original X = \n', X)
print()

print()
print('X with sorted columns :\n', np.sort(X, axis = 0))

print()
print('X with sorted rows :\n', np.sort(X, axis = 1))
```

### Arithmetic operations and Broadcasting

```python
x = np.array([1,2,3,4])
y = np.array([5.5,6.5,7.5,8.5])

print()
print('x = ', x)

print()
print('y = ', y)
print()

print('x + y = ', x + y)
print('add(x,y) = ', np.add(x,y))
print()
print('x - y = ', x - y)
print('subtract(x,y) = ', np.subtract(x,y))
print()
print('x  *y = ', x*  y)
print('multiply(x,y) = ', np.multiply(x,y))
print()
print('x / y = ', x / y)
print('divide(x,y) = ', np.divide(x,y))
```

2D examples

```python
X = np.array([1,2,3,4]).reshape(2,2)
Y = np.array([5.5,6.5,7.5,8.5]).reshape(2,2)

print()
print('X = \n', X)

print()
print('Y = \n', Y)
print()

print('X + Y = \n', X + Y)
print()
print('add(X,Y) = \n', np.add(X,Y))
print()
print('X - Y = \n', X - Y)
print()
print('subtract(X,Y) = \n', np.subtract(X,Y))
print()
print('X  *Y = \n', X*  Y)
print()
print('multiply(X,Y) = \n', np.multiply(X,Y))
print()
print('X / Y = \n', X / Y)
print()
print('divide(X,Y) = \n', np.divide(X,Y))
```

Other operations

```python
x = np.array([1,2,3,4])

print()
print('x = ', x)

print()
print('EXP(x) =', np.exp(x))
print()
print('SQRT(x) =',np.sqrt(x))
print()
print('POW(x,2) =',np.power(x,2))
```

```python
X = np.array([[1,2], [3,4]])

print()
print('X = \n', X)
print()

print('Average of all elements in X:', X.mean())
print('Average of all elements in the columns of X:', X.mean(axis=0))
print('Average of all elements in the rows of X:', X.mean(axis=1))
print()
print('Sum of all elements in X:', X.sum())
print('Sum of all elements in the columns of X:', X.sum(axis=0))
print('Sum of all elements in the rows of X:', X.sum(axis=1))
print()
print('Standard Deviation of all elements in X:', X.std())
print('Standard Deviation of all elements in the columns of X:', X.std(axis=0))
print('Standard Deviation of all elements in the rows of X:', X.std(axis=1))
print()
print('Median of all elements in X:', np.median(X))
print('Median of all elements in the columns of X:', np.median(X,axis=0))
print('Median of all elements in the rows of X:', np.median(X,axis=1))
print()
print('Maximum value of all elements in X:', X.max())
print('Maximum value of all elements in the columns of X:', X.max(axis=0))
print('Maximum value of all elements in the rows of X:', X.max(axis=1))
print()
print('Minimum value of all elements in X:', X.min())
print('Minimum value of all elements in the columns of X:', X.min(axis=0))
print('Minimum value of all elements in the rows of X:', X.min(axis=1))
```
