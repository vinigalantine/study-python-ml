# Learning Pandas

## Why should I use Pandas?

It is a package for data manipulation and analysis, providing DataFrame structures that make working with structured data fast and intuitive.

### Creating pandas Series

```python
import pandas as pd
import numpy as np
```

```python
groceries = pd.Series(data=[30,6,'Yes','No'], index=['eggs','apples','milk','bread'])

print(groceries)
print('\nShape:',groceries.shape)
print('\nRank:',groceries.ndim)
print('\nNum of values:',groceries.size)
print('\nIndices:',groceries.index)
print('\nValues:',groceries.ndim)
print('\nValues:',groceries.ndim)

print('\nIs bananas an index label in Groceries:', 'bananas' in groceries)
print('\nIs bread an index label in Groceries:', 'bread' in groceries)
```

### Accessing and Deleting Elements in pandas Series

```python
print("groceries['bread']:",groceries['bread'])
print("groceries.bread:",groceries.bread)

try:
    print(groceries['banana'])
    #OR
    print(groceries.banana)
except BaseException as e:
    print("Keys not found!")
    
print("\ngroceries['bread','eggs']:\n",groceries[['bread','eggs']])


print("\niloc",groceries.iloc[[0,2]])
```

Altering & Dropping

```python
print("Eggs before:",groceries['eggs'])

groceries['eggs'] = 20

print("\nEggs after:",groceries['eggs'])

print("\nBefore dropping:\n", groceries)
groceries.drop('apples')
print("\nAfter dropping:\n", groceries)

groceries.drop('apples', inplace=True)
print("\nAfter dropping with inplace:\n", groceries)
```

### Arithmetic Operations

```python
fruits= pd.Series(data = [10, 6, 3,], index = ['apples', 'oranges', 'bananas'])

print("Fruits:",fruits)

print('Original grocery list of fruits:\n ', fruits)

print()
print('fruits + 2:\n', fruits + 2)
print()
print('fruits - 2:\n', fruits - 2)
print()
print('fruits  *2:\n', fruits*  2)
print()
print('fruits / 2:\n', fruits / 2)
print()
```

You can use it with Numpy

```python
print('Original grocery list of fruits:\n', fruits)

print()
print('EXP(X) = \n', np.exp(fruits))
print() 
print('SQRT(X) =\n', np.sqrt(fruits))
print()
print('POW(X,2) =\n',np.power(fruits,2))
```

Performing operations on selected items

```python
print('Original grocery list of fruits:\n ', fruits)
print()

print('Amount of bananas + 2 = ', fruits['bananas'] + 2)
print()

print('Amount of apples - 2 = ', fruits.iloc[0] - 2)
print()

print('We double the amount of apples and oranges:\n', fruits[['apples', 'oranges']] * 2)
print()

print('We half the amount of apples and oranges:\n', fruits.loc[['apples', 'oranges']] / 2)
```

## Pandas DataFrames

```python
items = {'Alice': pd.Series(data = [40, 110, 500, 45], index = ['book', 'glasses', 'bike', 'pants']),
         'Bob': pd.Series(data = [245, 25, 55], index = ['bike', 'pants', 'watch'])}

print(type(items))

shopping_carts = pd.DataFrame(items)

print(shopping_carts)
```

Numerical rows

```python
data = {'Alice': pd.Series([40, 110, 500, 45]),
        'Bob': pd.Series([245, 25, 55])}

df = pd.DataFrame(data)

print(df)
```

Attributes

```python
print('shopping_carts has shape:', shopping_carts.shape)
print('shopping_carts has dimension:', shopping_carts.ndim)
print('shopping_carts has a total of:', shopping_carts.size, 'elements')
print()
print('The data in shopping_carts is:\n', shopping_carts.values)
print()
print('The row index in shopping_carts is:', shopping_carts.index)
print()
print('The column index in shopping_carts is:', shopping_carts.columns)
```

Creating with a specific column

```python
bob_shopping_cart = pd.DataFrame(items, columns=['Bob'])

print(bob_shopping_cart)
```

Selecting rows/columns of the DF

```python
sel_shopping_cart = pd.DataFrame(items, index = ['pants', 'book'])

print(sel_shopping_cart)

alice_sel_shopping_cart = pd.DataFrame(items, index = ['glasses', 'bike'], columns = ['Alice'])

print()
print(alice_sel_shopping_cart)
```

Creating a DF using a dict of lists

```python
data = {'Floats': [4.5, 8.2, 9.6],
        'Integers': [1, 2, 3]}

df = pd.DataFrame(data)

print(df)
```

Labeling the row-indexes

```python
data = {'Floats': [4.5, 8.2, 9.6],
        'Integers': [1, 2, 3]}

df = pd.DataFrame(data, index = ['label 1', 'label 2', 'label 3'])

print(df)
```

Using a list of dicts

```python
items2 = [{'bikes': 20, 'pants': 30, 'watches': 35}, 
          {'watches': 10, 'glasses': 50, 'bikes': 15, 'pants':5}]

store_items = pd.DataFrame(items2)

print(store_items)
```

### Accessing elements

```python
items2 = [{'bikes': 20, 'pants': 30, 'watches': 35}, 
          {'watches': 10, 'glasses': 50, 'bikes': 15, 'pants':5}]

store_items = pd.DataFrame(items2, index=['store 1', 'store 2'])

print(store_items)

print()
print('How many bikes are in each store:\n', store_items[['bikes']])
print()
print('How many bikes and pants are in each store:\n', store_items[['bikes', 'pants']])
print()
print('What items are in Store 1:\n', store_items.loc[['store 1']])
print()
print('How many bikes are in Store 2:', store_items['bikes']['store 2'])
```

Adding a column

```python
store_items['shirts'] = [15, 2]

print(store_items)
```

Adding a new column based on the arithmetic operation between existing columns of a DF

```python
store_items['suits'] = store_items['pants'] + store_items['shirts']

print(store_items)
```

Creating a row to be added to the DF and appending the row

```python
new_items = [{'bikes': 20, 'pants': 30, 'watches': 35, 'glasses': 4}]

new_store = pd.DataFrame(new_items, index = ['store 3'])

print(new_store)

store_items = pd.concat([store_items, new_store])

print(store_items)
```

Adding a new column that has data from existing columns

```python
store_items['new watches'] = store_items['watches'][1:]

print(store_items)
```

Adding a column to an index

```python
store_items.insert(5, 'shoes', [8, 5, 0])

print(store_items)
```

Deleting one or multiple column

```python
store_items.pop('new watches')

print(store_items)

store_items = store_items.drop(['watches', 'shoes'], axis=1)

print()
print(store_items)
```

Deleting rows

```python
store_items = store_items.drop(['store 2', 'store 1'], axis=0)

print(store_items)
```

Modifying the column and row labels

```python
store_items = store_items.rename(columns = {'bikes': 'hats'})

print(store_items)

store_items = store_items.rename(index = {'store 3': 'last store'})

print()
print(store_items)
```

Using existing column values as row-index

```python
store_items = store_items.set_index('pants')

print(store_items)
```

Dealing with NaN (Not a Number)

```python
items2 = [{'bikes': 20, 'pants': 30, 'watches': 35, 'shirts': 15, 'shoes':8, 'suits':45},
          {'watches': 10, 'glasses': 50, 'bikes': 15, 'pants':5, 'shirts': 2, 'shoes':5, 'suits':7},
          {'bikes': 20, 'pants': 30, 'watches': 35, 'glasses': 4, 'shoes':10}]

store_items = pd.DataFrame(items2, index = ['store 1', 'store 2', 'store 3'])

print("NaNs (True/False):",store_items.isnull())
print()
print("NaNs rows:",store_items.isnull().sum())
print("NaNs total:",store_items.isnull().sum().sum())

x =  store_items.isnull().sum().sum()

print()
print('Number of NaN values in our DataFrame:', x)
print()
print('Number of non-NaN values in the columns of our DataFrame:\n', store_items.count())
```

### Eliminating NaN Values

Drop rows/columns having NaN values

```python
print(store_items.dropna(axis = 0))
print(store_items.dropna(axis = 1))
```

### Substituting NaN values

Replacing NaN with 0

```python
print(store_items.fillna(0))
```

Foward fill NaN values down (axix=0) the dataframe

```python
print(store_items.ffill(axis=0))
```

Foward fill NaN values across (axix=0) the dataframe

```python
print(store_items.ffill(axis=1))
```

Backward fill NaN values down (axis=0) the dataframe

```python
print(store_items.bfill(axis=0))
```

Backward fill NaN values across (axis=1) the dataframe

```python
print(store_items.bfill(axis=1))
```

Interpolating (estimating) NaN values down (axis=0) the dataframe

```python
print(store_items.interpolate(method='linear', axis=0))
```

Interpolating (estimating) NaN values across (axis=0) the dataframe

```python
print(store_items.interpolate(method='linear', axis=1))
```
