# Learning PyTorch Basic Functions

## Key PyTorch Functions

````python
import torch
````

### Creating Random Tensors

- **torch.rand**: Creates random tensors.
- **Example**: `torch.rand(5)` creates a random vector with 5 elements.
- **Example**: `torch.rand(5, 5)` creates a 5x5 random tensor.

````python
vector = torch.rand(5)
print("Vector:", vector)

matrix = torch.rand(5, 5)
print("Matrix:\n", matrix)
````

### Creating Tensors Filled with Ones

- **torch.ones**: Creates tensors filled with ones.
- **Example**: `torch.ones(5)` creates a vector of size 5 filled with ones.
- **Example**: `torch.ones(5, 5)` creates a 5x5 tensor filled with ones.

````python
ones_vector = torch.ones(5)
print("Ones vector:", ones_vector)

ones_matrix = torch.ones(5, 5)
print("Ones matrix:\n", ones_matrix)
````

### Creating a Sequence of Numbers

- **torch.arange**: Creates a tensor with a sequence of numbers.
- **Example**: `torch.arange(10)` returns a sequence of numbers from 0 to 9.

````python
sequence = torch.arange(10)
print("Sequence:", sequence)
````

### Creating a Triangular Matrix

- **torch.tril**: Creates a lower triangular matrix.
- **Usage**: First, create a tensor, then apply `torch.tril` to set zeros above the main diagonal.

`torch.tril(torch.ones(5, 5))` keeps the main diagonal and everything below it:

| | col 0 | col 1 | col 2 | col 3 | col 4 |
|---|---|---|---|---|---|
| **row 0** | 1 | 0 | 0 | 0 | 0 |
| **row 1** | 1 | 1 | 0 | 0 | 0 |
| **row 2** | 1 | 1 | 1 | 0 | 0 |
| **row 3** | 1 | 1 | 1 | 1 | 0 |
| **row 4** | 1 | 1 | 1 | 1 | 1 |

> **Why this matters for Transformers:** read each row as "token *i* can see tokens 0 to *i*." Row 2 has ones only in columns 0–2, so token 2 can look at itself and earlier tokens, never future ones. This is the basis of the **causal mask** that stops a GPT-style model from peeking at the next word while learning to predict it.

````python
base = torch.ones(5, 5)
lower_triangular = torch.tril(base)
print("Lower triangular matrix:\n", lower_triangular)
````

### Concatenating Tensors

- **torch.cat**: Concatenates multiple tensors.
- **Usage**: Concatenate along rows by specifying `dim=0`.
- **Example**: `torch.cat((tensor1, tensor2), dim=0)`
- **Usage**: Concatenate along columns by specifying `dim=1`.
- **Example**: `torch.cat((tensor1, tensor2), dim=1)`

```mermaid
flowchart LR
    A["tensor1: shape (2, 3), all ones"] --> C{"torch.cat"}
    B["tensor2: shape (2, 3), all zeros"] --> C
    C -- "dim=0, extra rows" --> D["shape (4, 3)"]
    C -- "dim=1, extra columns" --> E["shape (2, 6)"]
```

The `dim` you pick is the one that grows. Every other dimension must match between the tensors.

````python
tensor1 = torch.ones(2, 3)
tensor2 = torch.zeros(2, 3)

rows_concat = torch.cat((tensor1, tensor2), dim=0)
print("Concatenated along rows (dim=0):\n", rows_concat)

cols_concat = torch.cat((tensor1, tensor2), dim=1)
print("Concatenated along columns (dim=1):\n", cols_concat)
````

### Reshaping Tensors

- **torch.view**: Changes the shape of a tensor.
- **Example**: `tensor.view(3, 4)` reshapes a vector into a 3x4 tensor.
- **Reverting Shape**: Use `view` to convert a 2D tensor back into a vector.
- **Example**: `tensor.view(12)` reshapes a 3x4 tensor into a vector of size 12.

```mermaid
flowchart LR
    A["torch.arange(12), shape (12,)"] -- ".view(3, 4)" --> B["shape (3, 4)"]
    B -- ".view(12)" --> C["shape (12,)"]
```

`view` only rearranges values. It never adds or removes any, so the element count has to stay the same: $3 \times 4 = 12$. Asking for `.view(5, 3)` on 12 elements raises an error.

````python
sequence = torch.arange(12)
reshaped = sequence.view(3, 4)
print("Reshaped to 3x4:\n", reshaped)

flattened = reshaped.view(12)
print("Flattened back to a vector:", flattened)
````
