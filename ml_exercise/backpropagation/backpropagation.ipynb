```python
import numpy as np
from helper import (
    sigmoid,
    grade_output_error,
    grade_output_error_term,
    grade_hidden_error_term,
    grade_delta_w_h_o,
    grade_delta_w_i_h
)

x = np.array([0.5, 0.1, -0.2])
target = 0.6
learnrate = 0.5

weights_input_hidden = np.array([[0.5, -0.6],
                                 [0.1, -0.2],
                                 [0.1,  0.7]])
weights_hidden_output = np.array([0.1, -0.3])

## Forward pass
hidden_layer_input  = np.dot(x, weights_input_hidden)
hidden_layer_output = sigmoid(hidden_layer_input)
output_layer_in     = np.dot(hidden_layer_output, weights_hidden_output)
output              = sigmoid(output_layer_in)
```

```python
## Backwards pass

# TODO: Calculate output error
# error = y - ŷ
error = target - output

grade_output_error(error, target, output)
```

```python
# TODO: Calculate error term for output layer
# h = ∑(w¹ × x¹)
# α = f(h) = sigmoid(h)
# ŷ = f(W × α) = sigmoid(W × α)
# f'(W × α) = f(W × α)(1-f(W × α))
# δ = (y - ŷ) × f'(W × α) 
output_error_term = error * output * ( 1 - output)

grade_output_error_term(output_error_term, error, output)
```

```python
# TODO: Calculate error term for hidden layer
# δ-h = W × δ × f'(h)

# weights_hidden_output: how much the hidden unit's result mattered to the final output
# output_error_term: how wrong the final output was (calculated above)
# hidden_layer_output * (1 - hidden_layer_output): how much the hidden unit's output could still change
# there's only 1 output here, so we just multiply everything together (no need to add anything up)
hidden_error_term = weights_hidden_output * output_error_term * hidden_layer_output * (1 - hidden_layer_output)

grade_hidden_error_term(hidden_error_term, output_error_term, weights_hidden_output, hidden_layer_output)
```

```python
# TODO: Calculate change in weights for hidden layer to output layer
# ΔW = ηδa

# learnrate: how big of a step to take
# output_error_term: how wrong the final output was (calculated above)
# hidden_layer_output: the hidden unit's actual output, which is what this weight receives
delta_w_h_o = learnrate * output_error_term * hidden_layer_output  # replace with your code

grade_delta_w_h_o(delta_w_h_o, learnrate, output_error_term, hidden_layer_output)
```

```python
# TODO: Calculate change in weights for input layer to hidden layer
# Δw = ηδ-hx

# learnrate: how big of a step to take
# x: the original input values that flowed into the hidden layer
# hidden_error_term: how wrong each hidden unit's output was (calculated above)
# np.outer pairs every input with every hidden error, building the (3, 2) grid of updates
delta_w_i_h = learnrate * np.outer(x, hidden_error_term)

print('Change in weights for hidden layer to output layer:')
print(delta_w_h_o)
print('Change in weights for input layer to hidden layer:')
print(delta_w_i_h)

grade_delta_w_i_h(delta_w_i_h, learnrate, hidden_error_term, x)
```
